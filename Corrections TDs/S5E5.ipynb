```python
# Reset de la mémoire
%reset -f
# Affichage dans des fenêtres indépendantes
%matplotlib qt5
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
# Fermer toutes les fenêtres graphiques
plt.close("all")
```

```python
# Lecture du fichier wav
# Le fichier mi2_bute_ros.wav doit être dans le même
# répertoire que le script python
Fe, note = wavfile.read("mi2_bute_ros.wav")
```

```python
N = len(note) # Nombre d'échantillons
df = Fe/N # Résolution fréquentielle
```

```python
## Construction des spectres bilatéraux
# Construction du vecteur fréquentiel
if N % 2 == 1 :
    # si N est impair
    f = np.arange(-Fe/2 + df/2,Fe/2 + df/2, df)
else :
    # Si N est pair
    f = np.arange(-Fe/2, Fe/2, df)
# Calcul de la transformée de Fourier bilatérale
# donc utilisation de fftshift
fftNote = np.fft.fftshift(np.fft.fft(note))/N;
# Calcul du module
ampfftNote = np.abs(fftNote);
# Calcul de la phase
phasefftNote = np.angle(fftNote);
## Affichage des spectres de module et de phase
# Affichage du module
plt.subplot(2,1,1);
plt.plot(f, ampfftNote);
plt.title('Spectre bilatéral de module de la note');
plt.xlabel('fréquence (Hz)');
plt.ylabel('Amplitude');
# Affichage de la phase
plt.subplot(2,1,2);
plt.plot(f, phasefftNote);
plt.title('Spectre bilatéral de phase de la note');
plt.xlabel('fréquence');
plt.ylabel('Phase (rad)');
plt.show()
figManager = plt.get_current_fig_manager()
figManager.window.showMaximized()
```

```python
plt.tight_layout()
```

```python
## Construction des spectres monolatéraux
# Construction du vecteur fréquentiel
if N % 2 == 1 :
    # si N est impair
    N2 = (N + 1)//2
    f = np.arange(0,Fe/2 + df/2, df)
else :
    # Si N est pair
    N2 = N//2
    f = np.arange(0, Fe/2, df)
# Calcul de la transformée de Fourier monolatérale
fftNote = np.fft.fft(note)/N;
# Calcul du module
ampfftNote = np.abs(fftNote);
# Manipulation du module pour monolatéral
ampfftNote = np.concatenate((ampfftNote[0], 2*ampfftNote[1:N2]), axis = None)
# Calcul de la phase
phasefftNote = np.angle(fftNote);
# Manipulation de la phase pour monolatéral
phasefftNote = phasefftNote[0:N2]
## Affichage des spectres de module et de phase
# Affichage du module
plt.figure() # Nouvelle figure
plt.subplot(2,1,1);
plt.plot(f, ampfftNote);
plt.title('Spectre monolatéral de module de la note');
plt.xlabel('fréquence (Hz)');
plt.ylabel('Amplitude');
# Affichage de la phase
plt.subplot(2,1,2);
plt.plot(f, phasefftNote);
plt.title('Spectre monolatéral de phase de la note');
plt.xlabel('fréquence');
plt.ylabel('Phase (rad)');
plt.show()
figManager = plt.get_current_fig_manager()
figManager.window.showMaximized()
```
